```python
import pandas as pd
import numpy as np
data={"name":["alice","bob","charlie","david","ema"],"age":[20,21,19,22,20],"marks":[85,72,np.nan,90,65],"attendance":[95,80,75,np.nan,88],"city":["kochi","delhi","mumbai","kochi","chennai"]}
df=pd.DataFrame(data)
print("orginal data frame")
print(df)
print("\nfirst 3 rows:")
print(df.head(3))

print("\nfirst 2 rows:")
print(df.tail(2))

print("\ndataframe shape")
print(df.shape)

print("\ndataframe columns")
print(df.columns)

print("\ndatatype")
print(df.dtypes)

print("\nmarks column:")
print(df["marks"])

print("\nName and marks:")
print(df[["name","marks"]])

print("\nfirst 3 rows:")
print(df.iloc[0:3])

print("\nstudents with marks greater than 75:")
print(df[df["marks"]>75])

print("\nstudents from kochi:")
print(df[df["city"]=="kochi"])

print("\nstudent with marks>70 AND attendence>80:")
print(df[
    (df["marks"]>70) &
    (df["attendance"]>80)
    ])

print("\nmarks +5:")
print(df["marks"]+5)

print("\narks *2:")
print(df["marks"]*2)

print("\naverage marks:")
print(df["marks"].mean())

print("\nmaximum marks:")
print(df["marks"].max())

print("\nminimum marks:")
print(df["marks"].min())

print("\ntotal marks:")
print(df["marks"].sum())

print("\nsorted by marks:")
print(df.sort_values("marks"))

df_renamed =df.rename(
    columns={
        "marks":"score",
        "attendance":"attendance_percentage"
    }
)

print("\nrenamed data frame:")
print(df_renamed)



```

```python
print("\nmissing value:")
print(df.isna)

print("\nnumber of missing value:")
print(df.isna().sum())
```

```python
df_forward =df.copy()
df_forward = df_forward.ffill()
print("\nforward fill:")
print(df_forward)
```

```python
df_backward =df.copy()
df_backward = df_backward.bfill()
print("\backward fill:")
print(df_backward)
```
