```python
import matplotlib.pyplot as plt

months=["jan","feb","mar","apr","may","jun","jul"]
sales=[20,30,25,40,50,80,10]
plt.plot(months,sales)


plt.xlabel("months")
plt.ylabel("sales")
plt.title("monthly sales")
plt.grid(True)
plt.plot(months,sales,marker="+",color="red",linestyle="-")
plt.show()
```
